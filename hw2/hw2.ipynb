### Common logic

In `common.py`, I implemented the `generate_motifs` method, which is the core logic used both in `q1.py` and `q2.py`.

High level explanation of the method:

**Input**: motif_size: Size of the motif to generate

**Output**: A list of all possible motifs of size `motif_size`.

1. Handles the corner cases for an empty motif and a singleton motif.
2. Creates a set of nodes numbered from 1 till `motif_size`.
3. Creates the set of all possible edges, by computing all permutations of 2 nodes.
4. Initializes 2 data structures which are used in order to optimize the runtime of the function. `id_to_motif_map` maps from a motif id to a triplet (motif_graph, sorted_in_degrees, sorted_out_degrees). `num_edges_to_motif_ids_map` maps from number of edges in the motif to a list of motif ids, such that all ids of motifs with the same number of edges will be listed together.
5. A motif must be weakly connected. Therefore, the minimal number of edges in it will be `motif_size`-1. Then, we can iterate over each `motif_size`-1 <= `num_edges` <= len(`possible_edges`):
- 5.1. Get all combinations of edges of size `num_edges`.
- 5.2. For each `edge_combination`:
  - 5.2.1. Construct a candidate motif, by initializing a directed graph over the `nodes` and `edge_combination` using `networkx` package.
  - 5.2.2. If the graph is not weakly connected, continue to the next `edge_combination`, as we know it can't be a motif candidate.
  - 5.2.3. Calculate the following invariants of the candidate motif - number of edges, and the sorted tuple of in degrees and out degrees.
  - 5.2.4. Try to find an already-discovered motif by using the `id_to_motif_map` and `num_edges_to_motif_ids_map` data structures. If a motif exists with the same invariants as the candidate motif (equal number of edges, and the sorted tuple of in degress and out degrees are equal element-wise):
    - 5.2.4.1. Perform the isomorphism test using VF2 algorithm, implemented in `networkx`.
  - 5.2.5. If the candidate motif is not isomorphic to any other already-discovered motif, add it's properties to the cache maps.


**Notes**:

- The isomorphism tests are crucial when considering motifs, as we don't care about node numbering/naming, and only care about the motif structure itself.
